```python

import random

equipment = {
    "attack": ["Gokon Tekko", "Shinten"],
    "spiritual_pressure": ["Fortification Pill", "Tenshintai"],
    "stamina": ["Cape", "Cookie"]
}


def randomizer_equipment():
    my_equipment = []

    while len(my_equipment) < 3:

        # Pick a random equipment category
        category = random.choice(list(equipment.keys()))

        # Pick a random item from that category
        euip_select = random.choice(equipment[category])

        # Prevent duplicate equipment
        if euip_select in my_equipment:
            continue

        confirmation_user = input(
            f"Randomly picked:\n{euip_select}. "
            "Do you have this equipment? (yes or no): "
        )

        if confirmation_user.lower() == "yes":
            my_equipment.append(euip_select)

            print(
                f"Great! {euip_select} added to your lineup. "
                f"{len(my_equipment)}/3"
            )
        else:
            print("No problem, picking different equipment...")

    return my_equipment


my_equipment = randomizer_equipment()

print(f"Your confirmed lineup is:\n{my_equipment}")
```
